**Exercice 2.1** Écrivez un script qui affiche, sur quatre sous-figures placées l'une en dessous de l'autre, la fonction $\sin(t)$ :

- au moyen de la fonction `plot` avec `t` variant de 0 à 4 avec 21 points ;
- au moyen de la fonction `plot` avec `t` variant de 0 à 4 avec 9 points ;

N'hésitez pas à vous aider de cette démo disponible sur le site de la librairie : https://matplotlib.org/stable/gallery/subplots_axes_and_figures/subplots_demo.html

```python
t1 = np.linspace(0, 4, 21)   # Permet de définir un vecteur composé de 21 points / éléments / nombres répartit de 0 à 4
t2 = np.linspace(0, 4, 9)

y1 = np.sin(t1)     # Calculer le Sinus de 't'
y2 = np.sin(t2)

fig, axs = plt.subplots(4, 1, figsize=(6, 8))   # Créer une figure avec 4 sous-figures (axes) l'une en dessous de l'autre


axs[0].plot(t1, y1, '-ro')
axs[0].set_title(f'Sous-figure 1')
axs[0].set_xlabel('t')
axs[0].set_ylabel('sin(t)')

axs[1].plot(t2, y2)
axs[1].set_title(f'Sous-figure 2')
axs[1].set_xlabel('t')
axs[1].set_ylabel('sin(t)')

axs[2].stem(t1, y1)
axs[2].set_title(f'Sous-figure 3')
axs[2].set_xlabel('t')
axs[2].set_ylabel('sin(t)')

axs[3].stem(t2, y2)
axs[3].set_title(f'Sous-figure 4')
axs[3].set_xlabel('t')
axs[3].set_ylabel('sin(t)')

plt.tight_layout()  # Ajuster les espaces entre les sous-figures

plt.show()  # Afficher la figure (contenant les 4 sous-figures)
```

*Dans l'ex ci-dessus on utilise des fonctions pour afficher des **Sinus** sous forme de Graphiques :*

- `t1 = np.linspace(0, 4, 21)` : Permet de définir un vecteur composé de 21 points / éléments / nombres répartit de 0 à 4.

- `y1 = np.sin(t1)` : permet de calculer le **Sinus** du vecteur `t`.

- `fig, axs = plt.subplots(4, 1, figsize=(6, 8))` : Permet de créer une figure avec 4 sous-figures (axes) l'une en dessous de l'autre. `4` et `1` permettent de définir le nombre de graphe en lignes et en colonnes. `figsize` permet de définir les dimensions de chaque graphe.

- `axs[0].plot(t1, y1)` : Permet de créer un graphe constitué d'une ligne reliant tous les points.

- `axs[2].stem(t1, y1)` : Permet de créer un graphe en nuage de points.

- `plt.tight_layout()` : Permet d'ajuster automatiquement les espaces entre les graphes pour pas qu'ils se chevauchent.

- `plt.show()` : Permet de faire le rendu des graphes / les faires apparaitre/afficher à l'écran.

**Exercice 2.6** Écrivez un code qui crée un signal sinusoïdal de 440 Hz, d'une amplitude de 0,5 et d'une durée de 4 s avec une fréquence d'échantillonnage de 22 kHz. <i>Indication</i> : si la fréquence d'échantillonnage est `fe`, le pas entre deux échantillons est `Te = 1/fe`. On aura alors `t = np.arange(0, 4 + Te, Te)`.

Écoutez (à l'aide de l'instruction `sd.play`) et visualisez ce signal. Pourquoi n'est-il pas possible de le visualiser correctement ?
Modifiez le signal pour que la visualisation soit possible, par exemple en se limitant à quatre périodes. Écoutez le nouveau signal. Pourquoi n'entendez-vous rien ?

Pour éviter de modifier le signal, vous pouvez aussi, lors de la visualisation, limiter l'intervalle de valeurs affichées en ajoutant `plt.xlim((0, 0.05))` avant `plt.show()` :

```python
# Un signal sinusoïdal est représenté par l'équation suivante : y(t)=Asin(2πft)

f = 440     # Fréquence du signal
A = 0.5     # Amplitude du signal
duration = 4    # Durée du signal en secondes
fe = 22000      # Fréquence déchantillonage

# Calcul du pas/de l'écart entre 2 échantillons qui est égal à '1/fe' avec 'fe' qui est la fréquence d'échantillonage.
Te = 1 / fe

# Création du temps 't' qui va de 0 à 4 par pas de 'Te'
t = np.arange(0, duration, Te)

# Création du signal Sinusoïdal
signal = A * np.sin(2 * np.pi * f * t)

# Lecture du signal
sd.play(signal, fe)

# Limitation de l'affichage de l'abscisse 'x' de 0 à 0,05 secondes
plt.xlim((0, 0.05))

# Visualisation du signal (signal complet, mais très dense)
plt.plot(t, signal)     # Représentation du signal avec une abscisse ayant comme échelle la matrice 't'
plt.title("Signal sinusoïdal complet (440 Hz)")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.show()
```

L'axe $x$ affichera uniquement le graphe pour $t$ allant de 0 à 0,05 s.

**Exercice 2.7** Écrivez un programme qui crée un signal qui est la somme d'un signal sinusoïdal de 440 Hz et d'un signal de 445 Hz, tous deux d'une durée de 4 s avec une fréquence d'échantillonnage de 22 kHz. Visualisez et écoutez le battement.

```python
f1 = 440  # Fréquence du premier signal (en Hz)
f2 = 445  # Fréquence du deuxième signal (en Hz)
A = 0.5   # Amplitude de chacun des signaux
duration = 4  # Durée du signal en secondes
fe = 22000  # Fréquence d'échantillonnage (en Hz)

# Calcul du pas/de l'écart entre 2 échantillons qui est égal à '1/fe' avec 'fe' qui est la fréquence d'échantillonage.
Te = 1 / fe

# Génération du temps 't' qui va de 0 à 4 par pas de 'Te'
t = np.arange(0, duration, Te)

# Création des deux signaux sinusoïdaux
signal1 = A * np.sin(2 * np.pi * f1 * t)
signal2 = A * np.sin(2 * np.pi * f2 * t)

# Signal résultant : somme des deux signaux
signal_sum = signal1 + signal2

# Lecture du signal
sd.play(signal_sum, fe)

# Visualisation du signal résultant
plt.plot(t, signal_sum)
plt.title("Signal résultant : Somme de 440 Hz et 445 Hz")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.xlim((0, 0.2))  # Limiter l'axe x pour mieux visualiser les oscillations
plt.show()
```

**Exercice 2.9** Sachant que le « La » de la troisième octave a une fréquence de 440 Hz et que l'on obtient la note suivante d'une gamme en multipliant la note précédente par $2^{1/6}$, créez une succession de 18 notes au moyen d'une boucle `for`. Choisissez une fréquence d'échantillonnage de 11025 Hz. Écoutez le résultat obtenu.

```python
f = 110     # Fréquence de la note de départ 
A = 0.5     # Amplitude du signal
n_notes = 18    # Nombres de notes
duration = 0.2      # Durée de chaque note
fe = 11025      # Fréquence d'échantillonage
Te = 1 / fe     # Ecart entre chaque échantillons

# Génération des 18 notes
notes = []
for i in range(n_notes):
    frequency = f * (2 ** (i/6))    # Calcul de la fréquence de la nouvelle note
    t =np.arange(0, duration, Te)   # Temps pour cette nouvelle note
    note = A * np.sin(2 * np.pi * frequency * t)    # Signal sinusoïdal pour la note
    notes.append(note)      # Ajout de la nouvelle note à la liste de notes

signal = np.concatenate(notes)      # Concaténation des notes pour en faire un seul signal

sd.play(signal, fe)
```

**Exercice 2.10** Créez un signal bruité de 440 Hz d'une durée de 4 s. Pour ajouter du bruit à votre signal sinusoïdal original, utilisez par exemple la fonction `np.random.rand`. Écoutez votre résultat.

```python
# Paramètres du signal
f = 440  # Fréquence du signal (Hz)
fe = 22000  # Fréquence d'échantillonnage (Hz)
duration = 4  # Durée du signal en secondes
A = 0.5  # Amplitude du signal sinusoïdal

# Génération du signal sinusoïdal
Te = 1 / fe  # Pas d'échantillonnage
t = np.arange(0, duration, Te)  # Temps
signal = A * np.sin(2 * np.pi * f * t)  # Signal sinusoïdal

# Génération du bruit aléatoire
noise = np.random.rand(len(t)) - 0.5  # Bruit centré autour de 0 (-0.5 à +0.5)

# Ajout du bruit au signal sinusoïdal
noisy_signal = signal + 0.1 * noise  # Le facteur 0.1 ajuste la force du bruit

# Écoute du signal bruité
sd.play(noisy_signal, fe)

# Visualisation du signal bruité (facultatif)
plt.plot(t[:1000], noisy_signal[:1000])  # On visualise les 1000 premiers points pour plus de lisibilité
plt.title("Signal sinusoïdal à 440 Hz avec du bruit")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.show()
```

**Exercice 2.11** Créez une oscillation amortie dont la fréquence est de 440 Hz et la durée est 2 s. Choisissez un facteur d'amortissement adéquat. Visualisez et écoutez le résultat.

```python
# Paramètres du signal
f = 440  # Fréquence du signal sinusoïdal (Hz)
fe = 22000  # Fréquence d'échantillonnage (Hz)
duration = 2  # Durée du signal (en secondes)
A = 0.5  # Amplitude initiale
alpha = 3  # Facteur d'amortissement

# Génération du temps
Te = 1 / fe  # Pas d'échantillonnage
t = np.arange(0, duration, Te)  # Tableau de temps

# Création du signal sinusoïdal amorti
signal = A * np.sin(2 * np.pi * f * t) * np.exp(-alpha * t)

# Visualisation du signal
plt.plot(t[:1000], signal[:1000])  # On visualise les 1000 premiers points pour mieux voir
plt.title("Oscillation amortie à 440 Hz")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")
plt.show()

# Lecture du signal
sd.play(signal, fe)
```

**Exercice 2.12** Écrivez un programme pour tester les signaux obtenus avec les fonctions suivantes et leurs différents paramètres : `np.sinc`, `signal.square`, `signal.sawtooth` et `signal.chirp`. Écoutez et affichez le résultat pour chacun d'eux.

```python
# Paramètres généraux
fe = 22000  # Fréquence d'échantillonnage (Hz)
duration = 2  # Durée des signaux (s)
t = np.linspace(0, duration, int(fe * duration), endpoint=False)  # Tableau de temps

# Fonction 1: np.sinc
# 2 * t - 1
sinc_signal = np.sinc(400 * t - 1)  # Décalé pour centrer sur 0. La fonction mathématique prise en param : sinc(x) = sin(pi * x) / (pi * x)

# Fonction 2: signal.square
square_signal = signal.square(2 * np.pi * 200 * t)  # Signal carré à 5 Hz. La formule mathématique prise en param : (2 * pi * hertz * temps)

# Fonction 3: signal.sawtooth
sawtooth_signal = signal.sawtooth(2 * np.pi * 200 * t)  # Signal en dents de scie à 5 Hz. La formule mathématique prise en param : (2 * pi * hertz * temps)

# Fonction 4: signal.chirp
# f1=fe/2
chirp_signal = signal.chirp(t, f0=10, f1=fe/10, t1=duration, method='linear')  # Chirp linéaire de 10 Hz à fe/2

# Visualisation des signaux
plt.figure(figsize=(10, 8))

plt.subplot(4, 1, 1)
plt.plot(t[:1000], sinc_signal[:1000])  # Affichage de 1000 premiers points
plt.title("Signal np.sinc")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")

plt.subplot(4, 1, 2)
plt.plot(t[:1000], square_signal[:1000])
plt.title("Signal carré (square)")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")

plt.subplot(4, 1, 3)
plt.plot(t[:1000], sawtooth_signal[:1000])
plt.title("Signal en dents de scie (sawtooth)")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")

plt.subplot(4, 1, 4)
plt.plot(t[:1000], chirp_signal[:1000])
plt.title("Signal Chirp")
plt.xlabel("Temps (s)")
plt.ylabel("Amplitude")

plt.tight_layout()
plt.show()

# Lecture des signaux
print("Lecture du signal sinc")
sd.play(sinc_signal, fe)
sd.wait()

print("Lecture du signal square")
sd.play(square_signal, fe)
sd.wait()

print("Lecture du signal sawtooth")
sd.play(sawtooth_signal, fe)
sd.wait()

print("Lecture du signal chirp")
sd.play(chirp_signal, fe)
sd.wait()
```
